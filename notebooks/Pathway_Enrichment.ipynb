# Pathway Enrichment Analysis

## TCGA-BRCA Molecular Subtypes

This notebook performs functional enrichment analysis on subtype-associated differentially expressed genes identified from TCGA-BRCA transcriptomic data.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import gseapy as gp
```

```python
#Load DEG Results
deg = pd.read_csv(
    "/home/sonia/bioinformatics/cancer-subtyping-multiomics/results/differential_expression_results.csv"
)

deg.head()
```

```python
#Filter Significant Genes
sig = deg[
    (deg["adj_pvalue"] < 0.05) &
    (abs(deg["logFC"]) > 1)
]

print("Significant genes:", sig.shape[0])

sig.head()
```

```python
#Extract Gene List
gene_list = sig["Gene"].dropna().tolist()

len(gene_list)
```

```python
#GO Biological Process Enrichment
enr_go = gp.enrichr(
    gene_list=gene_list,
    gene_sets="GO_Biological_Process_2023",
    organism="Homo sapiens",
    outdir=None
)

go_results = enr_go.results

go_results.head()
```

```python
#Top GO Terms
go_results[
    ["Term", "Adjusted P-value"]
].head(10)
```

```python
#GO Enrichment Plot
top_go = go_results.head(10)

plt.figure(figsize=(8,6))

plt.barh(
    top_go["Term"],
    -top_go["Adjusted P-value"].apply(lambda x: np.log10(x))
)

plt.xlabel("-log10 Adjusted P-value")
plt.title("Top GO Biological Processes")

plt.show()
```

```python
#KEGG Pathway Enrichment
enr_kegg = gp.enrichr(
    gene_list=gene_list,
    gene_sets="KEGG_2021_Human",
    organism="Human",
    outdir=None
)

kegg_results = enr_kegg.results

kegg_results.head()
```

```python
#Top KEGG Pathways
kegg_results[
    ["Term", "Adjusted P-value"]
].head(10)
```

```python
#KEGG Enrichment Plot
top_kegg = kegg_results.head(10)

plt.figure(figsize=(8,6))

plt.barh(
    top_kegg["Term"],
    -top_kegg["Adjusted P-value"].apply(lambda x: np.log10(x))
)

plt.xlabel("-log10 Adjusted P-value")
plt.title("Top KEGG Pathways")

plt.show()
```

```python
#Save Enrichment Results
go_results.to_csv(
    "/home/sonia/bioinformatics/cancer-subtyping-multiomics/results/go_enrichment.csv",
    index=False
)

kegg_results.to_csv(
    "/home/sonia/bioinformatics/cancer-subtyping-multiomics/results/kegg_enrichment.csv",
    index=False
)
```

# Interpretation

Pathway enrichment analysis identified biologically relevant signaling pathways associated with breast cancer molecular subtypes.

Enriched pathways included:

- Hormone receptor signaling
- Cell cycle regulation
- DNA replication
- Proliferation-associated pathways

These findings support the biological validity of transcriptomics-derived subtype classification and highlight subtype-specific molecular programs.
