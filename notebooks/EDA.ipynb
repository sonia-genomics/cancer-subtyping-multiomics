# Exploratory Data Analysis (EDA)

## TCGA-BRCA Transcriptomic Analysis

This notebook performs exploratory analysis of TCGA-BRCA RNA-seq expression data prior to downstream subtype identification and clustering.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load Expression Matrix
expr = pd.read_csv(
    "/home/sonia/bioinformatics/cancer-subtyping-multiomics/final_data/normalized_expression.csv",
    index_col=0
)

expr.head()
```

```python
#Dataset Shape
print("Expression matrix shape:", expr.shape)
```

```python
#Missing Values
print("Missing values:", expr.isnull().sum().sum())
```

```python
#Distribution of Expression Values
plt.figure(figsize=(8,5))

plt.hist(expr.values.flatten(), bins=100)

plt.xlabel("Normalized Expression")
plt.ylabel("Frequency")
plt.title("Distribution of Gene Expression Values")

plt.show()
```

```python
#PCA Exploration
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

pcs = pca.fit_transform(expr.T)

pca_df = pd.DataFrame(pcs, columns=["PC1", "PC2"])

pca_df.head()
```

```python
#PCA Plot
plt.figure(figsize=(7,6))

plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"],
    s=20
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA of TCGA-BRCA Samples")

plt.show()
```

```python
#Highly Variable Genes
gene_var = expr.var(axis=1)

top_var = gene_var.sort_values(ascending=False).head(20)

top_var
```

```python
#Heatmap of Top Variable Genes
top_genes = expr.loc[top_var.index]

plt.figure(figsize=(12,8))

sns.heatmap(top_genes)

plt.title("Top Variable Genes")

plt.show()
```

# Interpretation

EDA demonstrates substantial transcriptomic heterogeneity across TCGA-BRCA samples.

PCA indicates broad molecular variation among tumors, supporting downstream clustering analysis.

Highly variable genes may contribute to subtype-specific biological signatures and tumor heterogeneity.
